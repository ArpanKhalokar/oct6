```python
#append method
names = ["Arpan","Niraj","Yash"]
names.append("Vedant")
print(names)
```

```python
#insert method
names = ["Arpan","Niraj","Yash"]
names.insert(1, "Abhinay")
print(names)
```

```python
#extend method
names = ["Arpan","Niraj","Yash"]
names.extend(["Om","Nakul"])
print(names)
```

```python
#remove method
names = ["Arpan","Niraj","Yash"]
names.remove("Yash")
print(names)
```

```python
#pop method
names = ["Arpan","Niraj","Yash"]
names.pop(1)
print(names)
```

```python
#clear method
names = ["Arpan","Niraj","Yash"]
names.clear()
print(names)
```

```python
#index method
names = ["Arpan","Niraj","Yash"]
position = names.index("Yash")
print(position)
```

```python
#count method
number = [10,20,30,10,50,40]
print(number.count(10))
```

```python
#sort method
number = [10,20,30,10,50,40]
number.sort()
print(number)
```

```python
#reverse method
names = ["Arpan","Niraj","Yash"]
names.reverse()
print(names)
```

```python
#copy method
names = ["Arpan","Niraj","Yash"]
new_names = names.copy()
print(new_names)
```
